# 🚫 PART 2 — Stop Words

## What are Stop Words?

**Stop words** are very common words in a language that usually don't carry much meaning by themselves.

Examples in English: `is`, `the`, `a`, `an`, `he`, `she`, `are`, `was`...

When we analyze text, we often **remove** stop words so we can focus on the important words.

```
Original sentence:  "He is the best student in the class"
After removing stop words: ["best", "student", "class"]
```

# ✂️ PART 3 — Sentence Tokenization

## What is Tokenization?

**Tokenization** is the process of breaking text into smaller pieces called **tokens**.

- **Sentence Tokenization** → splits a paragraph into individual sentences
- **Word Tokenization** → splits a sentence into individual words

### Why is this tricky?

Not every `.` (period) ends a sentence!

```
"Dr. Smith works at I.B.M. He lives in the U.S.A."
```

Here, NLTK is smart enough to know which `.` ends a sentence and which does not.

````python
# ✅ Download required data for tokenization

nltk.download('punkt')
nltk.download('punkt_tab')
````

````python
# ✅ Sentence Tokenization using sent_tokenize()

from nltk.tokenize import sent_tokenize

# A paragraph about Dr. Narinder Singh Kapany (Father of Fiber Optics)
paragraph = (
    "Dr. Narinder Singh Kapany, widely known as the father of fiber optics, "
    "passed away peacefully in Woodside, California on December 3rd, 2020. "
    "In 1953, working alongside Harold Hopkins at Imperial College London, "
    "Dr. Kapany was the first to successfully transmit high quality images "
    "through fiber bundles. He coined the term fiber optics in a famous 1960 "
    "article for Scientific American. His breakthrough achievements helped "
    "create a foundational element of the internet age."
)

# Break the paragraph into sentences
sentences = sent_tokenize(paragraph)

print("Total characters in paragraph:", len(paragraph))
print("Number of sentences found:", len(sentences))
print()

# Print each sentence with its number
for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")
    print()
````

````python
# ✅ Notice how NLTK handles abbreviations like Mr., Dr., CEO

tricky_text = "Mr. Ram is the next CEO. He and Dr. Sita are ex students of AIML at IIT Ropar. He has great knowledge."

result = sent_tokenize(tricky_text)

print("Input text:")
print(tricky_text)
print()
print("Number of sentences:", len(result))
for i, s in enumerate(result, 1):
    print(f"  Sentence {i}: {s}")
````

# 🔤 PART 4 — Word Tokenization

**Word Tokenization** splits a sentence into individual words (and punctuation marks).

```
Input:  "I love NLP."
Output: ['I', 'love', 'NLP', '.']
```

Notice that punctuation like `.` and `,` are treated as separate tokens too!

````python
# ✅ Word Tokenization

from nltk.tokenize import word_tokenize

# Use the first sentence from the paragraph we tokenized earlier
first_sentence = sentences[0]
print("Sentence:")
print(first_sentence)
print()

# Break sentence into words
word_list = word_tokenize(first_sentence)

print("Number of tokens (words + punctuation):", len(word_list))
print()
print("Tokens:")
print(word_list)
````

````python
# ✅ Practice: Check if a word is a stop word
#
# .lower() converts a word to lowercase
# This is important because 'He' and 'he' should match the same stop word

test_words = ['He', 'vision', 'the', 'fiber', 'in']

print("Word             | Is Stop Word?")
print("-" * 30)

for word in test_words:
    is_stop = word.lower() in english_stopwords
    print(f"{word:<16} | {is_stop}")
````

# 🧹 PART 5 — Filtering Stop Words

Now let's combine what we learned:
1. Tokenize a sentence into words
2. Remove all stop words
3. Keep only the **meaningful** words

````python
# ✅ Method 1: Using a for loop to filter stop words
#
# We go through every word in our word_list.
# If it IS a stop word → skip it (pass)
# If it is NOT a stop word → keep it (append to result)

print("Original tokens:")
print(word_list)
print()

# Build a filtered list
filtered_words = []
for word in word_list:
    if word.lower() not in english_stopwords:
        filtered_words.append(word)

print("After removing stop words:")
print(filtered_words)
print()
print(f"Before: {len(word_list)} tokens  →  After: {len(filtered_words)} tokens")
````

````python
# ✅ Method 2: Shorter way using List Comprehension
#
# List comprehension is a compact way to write a for loop that builds a list.
# Format: [item for item in collection if condition]

filtered_words_v2 = [word for word in word_list if word.lower() not in english_stopwords]

print("Filtered words (using list comprehension):")
print(filtered_words_v2)
````

````python
# ✅ Practice: Tokenize a new sentence and check each word

sample = "Sita Ram is not a good halwai."
tokens = word_tokenize(sample)

print(f"Sentence: '{sample}'")
print()
print("Word             | Stop Word?")
print("-" * 30)

for word in tokens:
    if word.lower() in english_stopwords:
        status = "YES (stop word)"
    else:
        status = "no"
    print(f"{word:<16} | {status}")
````

````python
# ✅ Scrape a webpage and extract all text
#
# urllib.request.Request() lets us add a 'User-Agent' header.
# Some websites block bots; adding a browser User-Agent helps us look like a real browser.

import urllib.request
from bs4 import BeautifulSoup

url = "https://www.geeksforgeeks.org/machine-learning/machine-learning/"

# Pretend to be a browser so the website doesn't block us
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
}
request = urllib.request.Request(url, headers=headers)

# Download the page
response = urllib.request.urlopen(request)
raw_html = response.read()

# Parse with BeautifulSoup
soup = BeautifulSoup(raw_html, "html5lib")

# Extract all text
page_text = soup.get_text(strip=True)

print("Page scraped successfully!")
print(f"Total characters in text: {len(page_text)}")
print()
print("First 300 characters:")
print(page_text[:300])
````

````python
# ✅ Search for specific content in the scraped text
#
# .split() breaks a string into a list of words, splitting on spaces.

word_tokens = page_text.split()
print(f"Total words on page: {len(word_tokens)}")
print("First 20 words:", word_tokens[:20])

print()

# Search for a keyword
search_word = "machine"
found = [w for w in word_tokens if search_word.lower() in w.lower()]
print(f"Words containing '{search_word}': {len(found)}")
print("Examples:", found[:10])
````

# 📊 PART 7 — Word Frequency with FreqDist

**Frequency Distribution** tells us how many times each word appears in a text.

This is useful to find out what a text is *about* — the most frequent meaningful words are usually the key topics.

````python
# ✅ Count word frequencies using NLTK's FreqDist

# Build a frequency distribution from the page words
freq_dist = nltk.FreqDist(word_tokens)

print("Type:", type(freq_dist))
print()

# Show words that appear more than 5 times
print("Words appearing more than 5 times:")
print(f"{'Word':<20} | Count")
print("-" * 30)

for word, count in freq_dist.most_common(20):
    if count >= 5:
        print(f"{word:<20} | {count}")
````

````python
# ✅ Reusable function: showFreq(url)
#
# A function is a block of reusable code.
# def keyword defines a function.
# We pass a URL as input and it prints frequent words from that page.

import ssl

def showFreq(url, min_count=3):
    """
    Opens a URL and prints all words that appear at least min_count times.
    Parameters:
        url       : the web page to scrape
        min_count : minimum frequency to print (default: 3)
    """
    # Bypass SSL certificate errors (some sites have self-signed certs)
    context = ssl._create_unverified_context()

    response = urllib.request.urlopen(url, context=context)
    raw_html = response.read()

    soup = BeautifulSoup(raw_html, "html5lib")
    text = soup.get_text(strip=True)

    words = text.split()
    fd = nltk.FreqDist(words)

    print(f"Frequent words (appearing >= {min_count} times) on: {url}")
    print()
    for word, count in fd.most_common(15):
        if count >= min_count:
            print(f"  {word:<25} — {count}")


# Try it on a Python tutorial page
showFreq("https://www.w3schools.com/python/python_getstarted.asp", min_count=5)
````

````python
# ✅ Reusable function: Count how many times a specific word appears on a page

def count_word_on_page(url, target_word):
    """
    Counts how many times target_word appears on the given URL.
    Case-insensitive search.
    """
    response = urllib.request.urlopen(url)
    raw_html = response.read()
    soup = BeautifulSoup(raw_html, "html5lib")
    text = soup.get_text(strip=True)

    words = [w.lower() for w in text.split()]
    fd = nltk.FreqDist(words)

    count = fd[target_word.lower()]
    print(f"The word '{target_word}' appears {count} time(s) on the page.")


count_word_on_page("https://www.w3schools.com/python/python_getstarted.asp", "python")
````

# 📚 PART 8 — WordNet: Synonyms, Antonyms & Definitions

## What is WordNet?

**WordNet** is a large English dictionary database built into NLTK. For any English word, WordNet gives you:

- **Synsets** → groups of words with the same meaning (synonym sets)
- **Definitions** → meaning of each sense
- **Lemmas** → different forms of the word
- **Antonyms** → opposite words

One word can have **multiple meanings** (called *senses*). For example, the word *bank* can mean:
1. A financial institution
2. The side of a river
3. To tilt while turning (airplane)

````python
# ✅ Download WordNet data

nltk.download('wordnet')
````

````python
# ✅ Explore multiple meanings (synsets) of a word

from nltk.corpus import wordnet

def show_meanings(word):
    """Prints all meanings (synsets) of a given word."""
    synsets = wordnet.synsets(word)
    print(f"Word: '{word}' has {len(synsets)} meanings:")
    print()
    for i, syn in enumerate(synsets[:6], start=1):
        print(f"  Meaning {i} ({syn.name()}):")
        print(f"    Definition: {syn.definition()}")
        print()

show_meanings("bank")
````

````python
# ✅ Get synonyms of a word
#
# syn.lemmas() gives all word forms in that synonym group
# lemma.name() gives the actual word string

def get_synonyms(word):
    """Returns all unique synonyms of a word."""
    synonyms = set()  # Using a set to avoid duplicates
    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name().replace('_', ' '))
    return synonyms

print("Synonyms of 'good':")
print(get_synonyms("good"))

print()
print("Synonyms of 'happy':")
print(get_synonyms("happy"))
````

````python
# ✅ Get antonyms (opposite words)
#
# lemma.antonyms() returns a list of antonym lemmas
# If a lemma has no antonym, the list is empty

def show_antonyms(word):
    """Prints words and their antonyms (if available)."""
    print(f"Antonyms related to '{word}':")
    print()
    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            if lemma.antonyms():
                print(f"  {lemma.name()}  ↔  {lemma.antonyms()[0].name()}")

show_antonyms("good")
````

# 🌱 PART 9 — Stemming & Lemmatization

## The Problem: Many Forms of the Same Word

The words `runs`, `running`, `ran`, and `runner` all come from the root word `run`.

In NLP, we often want to treat all of these as the same word. Two techniques help:

| Technique | What it does | Example |
|-----------|-------------|--------|
| **Stemming** | Chops off word endings (fast but rough) | `playing` → `play`, `classes` → `class` |
| **Lemmatization** | Finds the proper base form (slower but accurate) | `playing` → `play`, `better` → `good` |

> 💡 Stemming can produce non-real words (e.g., `studies` → `studi`). Lemmatization always gives a real word.

````python
# ✅ Stemming with PorterStemmer
#
# Porter Stemmer applies rules to strip suffixes from words.
# It is fast but sometimes produces non-real words.

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

test_words = ['heros', 'classes', 'buses', 'increases', 'playing', 'studies', 'running', 'happily']

print("Stemming Results:")
print(f"{'Original':<15} | {'Stemmed':<15}")
print("-" * 35)

for word in test_words:
    print(f"{word:<15} | {stemmer.stem(word):<15}")
````

````python
# ✅ Lemmatization with WordNetLemmatizer
#
# Lemmatization uses a dictionary to find the proper root form.
# You can specify the Part of Speech (pos):
#   pos='v'  → treat word as a verb
#   pos='n'  → treat word as a noun
#   pos='a'  → treat word as an adjective
#   pos='r'  → treat word as an adverb

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Compare results with different POS tags
word = "playing"
print(f"Lemmatizing the word: '{word}'")
print()
print(f"  As a verb (pos='v'): {lemmatizer.lemmatize(word, pos='v')}")
print(f"  As a noun (pos='n'): {lemmatizer.lemmatize(word, pos='n')}")
print(f"  As adjective (pos='a'): {lemmatizer.lemmatize(word, pos='a')}")
````

````python
# ✅ Compare Stemming vs Lemmatization side by side

words_to_compare = ['increases', 'buses', 'classes', 'running', 'better', 'studies']

print(f"{'Word':<12} | {'Stemmed':<12} | {'Lemmatized (verb)'}")
print("-" * 45)

for word in words_to_compare:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word, pos='v')
    print(f"{word:<12} | {stemmed:<12} | {lemmatized}")
````

````python
# ✅ Reusable function: Show all lemma forms of a word

def show_lemma_forms(word):
    """Shows how a word is lemmatized for each part of speech."""
    print(f"Lemma forms of: '{word}'")
    print(f"  Verb      : {lemmatizer.lemmatize(word, pos='v')}")
    print(f"  Noun      : {lemmatizer.lemmatize(word, pos='n')}")
    print(f"  Adjective : {lemmatizer.lemmatize(word, pos='a')}")
    print(f"  Adverb    : {lemmatizer.lemmatize(word, pos='r')}")

show_lemma_forms('plays')
print()
show_lemma_forms('better')
````

# 🤔 PART 10 — Sentiment Analysis (Your Own Code!)

## What is Sentiment Analysis?

**Sentiment Analysis** is the task of figuring out whether a piece of text expresses a **positive** or **negative** feeling.

Examples:
- `"Great food and amazing service!"` → **Positive** ✅
- `"The place was dirty and service was slow."` → **Negative** ❌

## Our Approach: Naive Bayes Classifier

We will build a simple classifier using NLTK's **Naive Bayes** algorithm. The steps are:

1. **Training data** — sentences labeled pos/neg
2. **Dictionary** — all unique words across all training sentences
3. **Feature extraction** — for each sentence, record which words from the dictionary are present
4. **Train** the classifier
5. **Predict** sentiment of new sentences

````python
# ✅ Step 1: Prepare labeled training data
#
# Each item is a tuple: (sentence, label)
# 'pos' = positive sentiment, 'neg' = negative sentiment

train = [
    ("Great place to be when you are in Bangalore.", "pos"),
    ("The place was being renovated when I visited so the seating was limited.", "neg"),
    ("Loved the ambience, loved the food", "pos"),
    ("The food is delicious but not over the top.", "neg"),
    ("Service was a little slow, probably because too many people.", "neg"),
    ("The place is not easy to locate", "neg"),
    ("Mushroom fried rice was spicy", "pos"),
    ("Mushroom fried rice was good.", "pos"),
    ("Burfi is very sweet.", "pos"),
    ("Burfi is not sweet.", "neg"),
]

print("Training data:")
for sentence, label in train:
    print(f"  [{label.upper()}] {sentence}")
````

````python
# ✅ Step 2: Build a dictionary of all unique words
#
# We go through every sentence, tokenize it, lowercase every word,
# and collect all unique words into a set.
#
# set() automatically removes duplicate words.

dictionary = set(
    word.lower()
    for sentence, label in train
    for word in word_tokenize(sentence)
)

print(f"Total unique words in training data: {len(dictionary)}")
print()
print("Sample words:", list(dictionary)[:15])
````

````python
# ✅ Step 3: Extract features from each training sentence
#
# For every sentence, we create a dictionary like:
#   {'great': True, 'place': True, 'not': False, 'fried': False, ...}
# This tells the classifier which words from our vocabulary appear in each sentence.

def extract_features(sentence, dictionary):
    """Returns a feature dict: True for each word that appears in sentence."""
    words_in_sentence = set(word.lower() for word in word_tokenize(sentence))
    return {word: (word in words_in_sentence) for word in dictionary}

# Build training features
training_features = [
    (extract_features(sentence, dictionary), label)
    for sentence, label in train
]

print(f"Number of training examples: {len(training_features)}")
print()
# Show features that are True for the first sentence
first_features = training_features[0][0]
print(f"Features present in: '{train[0][0]}'")
present = {k: v for k, v in first_features.items() if v}
print(present)
````

````python
# ✅ Step 4: Train the Naive Bayes Classifier
#
# We pass the (features, label) pairs to the classifier.
# It learns which words are associated with 'pos' vs 'neg'.

classifier = nltk.NaiveBayesClassifier.train(training_features)

print("Classifier trained successfully!")
print()
print("Most informative features (words that help classify best):")
classifier.show_most_informative_features(10)
````

````python
# ✅ Step 5: Predict sentiment of new sentences

def predict_sentiment(text, classifier, dictionary):
    """
    Predicts whether text is 'pos' (positive) or 'neg' (negative).
    """
    features = extract_features(text, dictionary)
    prediction = classifier.classify(features)
    emoji = "✅ Positive" if prediction == "pos" else "❌ Negative"
    print(f"Text: '{text}'")
    print(f"Sentiment: {emoji}")
    print()

# Test with new sentences
predict_sentiment("Mushroom fried rice was good.", classifier, dictionary)
predict_sentiment("Burfi is not sweet.", classifier, dictionary)
predict_sentiment("Burfi is sweet.", classifier, dictionary)
predict_sentiment("Great Place is getting renovated.", classifier, dictionary)
````

# 📝 PART 12 — Extractive Text Summarization

## What is Text Summarization?

**Summarization** means automatically creating a shorter version of a long text that preserves the key ideas.

There are two types:

| Type | How it works | Example |
|------|-------------|--------|
| **Extractive** | Picks the most important *existing* sentences | Copies the 2 best sentences |
| **Abstractive** | Generates *new* sentences (like a human would) | Rewrites in shorter form |

We will use **Extractive Summarization** with the **TextRank** algorithm — which works like Google's PageRank but for sentences.

````python
# ✅ TextRank Summarization
#
# Algorithm Steps:
#   1. Split text into sentences
#   2. Convert sentences to TF-IDF vectors (numbers representing the sentence)
#   3. Compute similarity between every pair of sentences
#   4. Build a graph where sentences are nodes and similarity = edge weight
#   5. Apply PageRank to find the most 'central' (important) sentences
#   6. Return top N sentences as the summary

import networkx as nx
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def extractive_summary(text, num_sentences=2):
    """
    Summarizes the input text by picking the most important sentences.
    Parameters:
        text          : the article or paragraph to summarize
        num_sentences : how many sentences to include in summary
    """
    # Step 1: Split into sentences
    sentences = sent_tokenize(text)

    if len(sentences) <= num_sentences:
        print("Text is already short enough!")
        return text

    # Step 2: Convert sentences to TF-IDF vectors
    vectorizer = TfidfVectorizer(stop_words='english')
    sentence_vectors = vectorizer.fit_transform(sentences)

    # Step 3: Calculate similarity between all sentence pairs
    similarity_matrix = cosine_similarity(sentence_vectors)

    # Step 4: Build graph and apply PageRank
    graph = nx.from_numpy_array(similarity_matrix)
    scores = nx.pagerank(graph)

    # Step 5: Rank and select top sentences
    ranked = sorted(
        ((scores[i], s) for i, s in enumerate(sentences)),
        reverse=True
    )

    print(f"--- Original Text ({len(sentences)} sentences) ---")
    print(text.strip())
    print()
    print(f"--- Summary ({num_sentences} most important sentences) ---")
    for score, sentence in ranked[:num_sentences]:
        print(f"  [Score: {score:.4f}] {sentence}")


# Test with a sample article about NLP
article = """
Natural language processing (NLP) is a subfield of linguistics, computer science, and artificial intelligence.
It focuses on the interactions between computers and human language.
The goal is to strictly process and analyze large amounts of natural language data.
The result is a computer capable of understanding the contents of documents, including the contextual nuances of the language within them.
The technology can then accurately extract information and insights contained in the documents, as well as categorize and organize the documents themselves.
"""

extractive_summary(article, num_sentences=2)
````

# 🎯 BONUS — Word Ambiguity: One Word, Many Meanings

NLP is hard because human language is **ambiguous**.

**Lexical Ambiguity**: One word has multiple meanings
- `"I went to the bank"` → bank of a river? or a financial bank?

**Structural Ambiguity**: A sentence can be parsed in multiple ways
- `"I saw the man with the telescope."`
  - Did **I** use the telescope to see the man?
  - Or did **the man** have the telescope?

This is why NLP is an interesting challenge — and why modern AI models (like ChatGPT) are so impressive for handling it!
